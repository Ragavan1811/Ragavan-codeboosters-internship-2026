```python
import pandas as pd
df=pd.read_csv("student_performance.csv")
print("python library")
```

```python
electronics=df[df['department']=='Electronics']     #1
print("student from electronic department:",len(electronics))
```

```python
female_attendance = df[df["gender"] == "Female"]["attendance_percentage"].mean()    #2

print("Average attendance of female students :", female_attendance)
```

```python
lowest_programming = df.loc[df["programming_score"].idxmin()]       #3

print("Student with lowest programming score :", lowest_programming["name"])
```

```python
above_80_math = len(df[df["math_score"] > 80])         #4

print("Students scored above 80 in math :", above_80_math)
```

```python
total_score=df['english_score']+df['math_score']+df['programming_score']     #5
df['total_score']

avg_total_above_90 = df[df["attendance_percentage"] > 90]["total_score"].mean()
print("Average total score :", avg_total_above_90)
```
